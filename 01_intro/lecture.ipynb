> **Вывод.** Наименьшее число вопросов, которых гарантированно хватит.
> **Ограничения.** 1 секунда, 64 МБ.

Число делений пополам, за которое `n` вариантов сжимаются до одного, называют **двоичным логарифмом** (binary logarithm, `log2`). Для `1000` это чуть меньше `10`, и последний, неполный шаг тоже стоит вопроса. Поэтому логарифм округляют вверх. На занятии округление вверх написали руками, через прибавку, а не через `math.ceil`:

````python
import math

n = int(input())
print(int(math.log2(n) + 0.9999999999))
````

<details><summary>Ответ</summary>

Прибавка даёт `11.0014`, и `int` оставляет `11`. То же самое делает `math.ceil`. У прибавки есть слепая зона: если дробная часть логарифма меньше `10^-10`, округления вверх не будет. В пределах задачи, `N ≤ 10^5`, таких `N` нет, и код проходит.

Стратегия «отбросить половину» называется **бинарный поиск** (binary search). Число шагов растёт как логарифм: удвоение `n` добавляет один шаг. Миллион вариантов это 20 вопросов, миллиард это 30.

## 2. Роль первая: поиск в отсортированном массиве

Есть отсортированный массив из 14 чисел. Нужно узнать, есть ли в нём число `4`, и если есть, то где оно встречается впервые.

````python
arr = [1, 1, 2, 2, 2, 3, 3, 4, 4, 4, 4, 7, 8, 9]
````

Наивно: `arr.index(4)` идёт слева направо и делает 8 сравнений. Для числа, которого нет, все 14. На массиве в миллион элементов и миллионе запросов это миллион запросов по миллиону сравнений, `10^12`. `arr.index` перебирает около 150 миллионов элементов в секунду: `10^12 / (1.5 · 10^8) ≈ 7000` секунд, около двух часов. А ведь массив отсортирован, и этот порядок никак не используется.

Проверьте на своей машине: тысяча запросов к списку из миллиона чисел.

````python
import time

big = list(range(10**6))
start = time.perf_counter()
for _ in range(1000):
    big.index(10**6 - 1)          # худший случай: искомое в самом конце
print(time.perf_counter() - start)
````

Печатает около `6` секунд. Миллион запросов это в тысячу раз дольше. Теперь то же самое бинарным поиском.

### Первое вхождение: код с занятия по шагам

Вот код с занятия. Подписи справа называют назначение каждого шага.

````python
def bin_first(key: int, arr: list[int]) -> int:
    left, right = -1, len(arr)                    # 1. ГРАНИЦЫ: arr[left] < key <= arr[right]; за краями массива считаем это верным
    while right - left > 1:                       # 2. ПОКА между границами есть хотя бы один индекс
        mid = (left + right) // 2                 # 3. СЕРЕДИНА
        if arr[mid] < key:                        # 4. ПРОВЕРКА: середина ещё левее ответа?
            left = mid                            #    да: левая граница прыгает в середину
        else:
            right = mid                           #    нет: правая граница прыгает в середину
    if right != len(arr) and arr[right] == key:   # 5. ОТВЕТ: right стоит на первом элементе >= key
        return right
    else:
        return -1


print(bin_first(4, arr))
````

**Предскажите.** Сколько раз выполнится цикл для `key = 4`, и на каких индексах остановятся `left` и `right`?

---

<details><summary>Ответ и анимация</summary>

Четыре раза. `left` остановится на `6`, `right` на `7`. Синим закрашено то, что уже точно меньше `key`, оранжевым то, что точно не меньше. Каждый шаг отдаёт одной из зон половину оставшегося, пока зоны не сомкнутся. Ячейки `-1` и `14` пунктирные: их в массиве нет, это границы за краями.

Таблица шагов (трасса, trace) вызова `bin_first(4, arr)`:

| шаг | left | right | mid | arr[mid] | действие |
|---|---|---|---|---|---|
| 1 | -1 | 14 | 6 | 3 | `3 < 4`, `left = 6` |
| 2 | 6 | 14 | 10 | 4 | `right = 10` |
| 3 | 6 | 10 | 8 | 4 | `right = 8` |
| 4 | 6 | 8 | 7 | 4 | `right = 7` |
| стоп | 6 | 7 | | | `right - left == 1` |

`arr[7] == 4`, ответ `7`. Четыре шага вместо восьми, а на миллионе элементов двадцать вместо миллиона.

Посмотрите на утверждение из шага 1: `arr[left] < key <= arr[right]`. Оно верно перед циклом и после каждого его шага. В трассе `left` останавливается на `6`, где стоит тройка, а `right` на `7`, где стоит первая четвёрка. Такое утверждение, которое цикл сохраняет на каждом шаге, называют **инвариантом** (invariant). Он и есть смысл кода: `left` всегда левее ответа, `right` на ответе или правее.

Начальные границы `-1` и `len(arr)` лежат за пределами массива. Так инвариант верен с самого начала: левее `-1` нет элементов, правее `len(arr)` тоже. По этим индексам код никогда не читает массив. В Python `arr[-1]` это последний элемент, но `mid` до `-1` не доходит: пока цикл работает, `right` больше `left` хотя бы на два, значит, `mid` не меньше `0`. По той же причине `mid` не доходит и до `len(arr)`. Когда границы становятся соседями, `right` указывает на первый элемент, который не меньше `key`.

### Последнее вхождение: задача из контеста

На занятии этим вариантом решали задачу из контеста.

> **Задача C. Поиск.** Реализуйте алгоритм бинарного поиска.
> **Ввод.** В первой строке натуральные числа `N` и `K` (`1 ≤ N, K ≤ 1000000`). Во второй строке `N` элементов первого массива, отсортированного по возрастанию. В третьей строке `K` элементов второго массива. Элементы обоих массивов целые, по модулю не больше `10^9`.
> **Вывод.** Для каждого из `K` чисел вывести в отдельную строку `YES`, если это число встречается в первом массиве, и `NO` в противном случае.
> **Ограничения.** 5 секунд, 256 МБ.
>
> Пример 1, ввод:
> ```
> 3 3
> 2 3 5
> 1 2 3
> ```
> Вывод: `NO`, `YES`, `YES`.
>
> Пример 2, ввод:
> ```
> 3 4
> 1 2 2
> 1 2 4 5
> ```
> Вывод: `YES`, `YES`, `NO`, `NO`.

Пример 2 уточняет слова «по возрастанию»: повторы в массиве допустимы, `1 2 2`. Ограничения `N, K ≤ 10^6` это ровно тот миллион запросов по миллиону элементов, с которого начался раздел. В коде с занятия `K` назван `m`.

````python
# введите три строки примера 1: «3 3», «2 3 5», «1 2 3»; ответ: NO, YES, YES
n, m = map(int, input().split())

arr = list(map(int, input().split()))


def bin_last(key: int, arr: list[int]) -> int:
    left, right = -1, len(arr)                    # 1. ГРАНИЦЫ: arr[left] <= key < arr[right]; за краями считаем верным
    while right - left > 1:                       # 2. ПОКА между границами есть индекс
        mid = (left + right) // 2                 # 3. СЕРЕДИНА
        if arr[mid] <= key:                       # 4. ПРОВЕРКА: середина не правее ответа?
            left = mid
        else:
            right = mid
    if left != -1 and arr[left] == key:           # 5. ОТВЕТ: left стоит на последнем элементе <= key
        return left
    else:
        return -1


for key in map(int, input().split()):
    if bin_last(key, arr) != -1:
        print("YES")
    else:
        print("NO")
````

`n` и `m` прочитаны, но дальше не используются: `split()` сам разбирает строку целиком.

Трасса `bin_last(4, arr)` на том же массиве из начала раздела:

| шаг | left | right | mid | arr[mid] | действие |
|---|---|---|---|---|---|
| 1 | -1 | 14 | 6 | 3 | `3 <= 4`, `left = 6` |
| 2 | 6 | 14 | 10 | 4 | `4 <= 4`, `left = 10` |
| 3 | 10 | 14 | 12 | 8 | `right = 12` |
| 4 | 10 | 12 | 11 | 7 | `right = 11` |
| стоп | 10 | 11 | | | `arr[10] == 4`, ответ `10` |

Сравните две функции:

| | `bin_first` | `bin_last` |
|---|---|---|
| инвариант | `arr[left] < key <= arr[right]` | `arr[left] <= key < arr[right]` |
| проверка середины | `arr[mid] < key` | `arr[mid] <= key` |
| какая граница держит ответ | `right` | `left` |
| защита от выхода за край | `right != len(arr)` | `left != -1` |
| границы, цикл, середина | одинаковые | одинаковые |

Общая форма этого принципа: бинарный поиск ищет границу между «ещё нет» и «уже да» на упорядоченной шкале. Скелет всегда один. Меняется только вопрос, который задают середине, и то, какая из двух границ в конце указывает на ответ.

## 3. Зачем это учить: та же роль в базе данных

Пока приём выглядит как трюк на массиве из 14 чисел. Вот место, где он стоит денег, и где вы почти наверняка уже с ним сталкивались.

Таблица клиентов на десять миллионов строк. Запрос «найти клиента по номеру» приходит тысячу раз в секунду. Без подготовки каждый запрос просматривает все строки: десять миллионов строк на тысячу запросов, `10^10` чтений в секунду. Быстрый SSD делает около миллиона чтений в секунду, нужно в десять тысяч раз больше.

Вы уже знаете, что делать. Если номера клиентов лежат в отсортированном виде, поиск по ним это 24 шага вместо десяти миллионов: `log2(10^7) ≈ 23.3`. База данных строит такую отсортированную структуру рядом с таблицей и называет её **индексом** (index).

**Предскажите.** Таблица выросла в тысячу раз. Во сколько раз вырос поиск без индекса? А на сколько шагов вырос поиск по индексу?

---

<details><summary>Ответ</summary>

Без индекса ровно в тысячу раз: просматриваются все строки. По индексу примерно на десять шагов: `log2(1000) ≈ 10`, было 24, стало 34.
</details>

---

Вот и ответ на вопрос, зачем учить алгоритмы тому, кто их не пишет. Индекс руками не делает никто, база строит его одной командой. Но пойдёт запрос десять миллисекунд или четыре минуты, решает одно: ложится ли он на упорядоченную шкалу. Сравните два запроса к одной и той же таблице с одним и тем же индексом по номеру телефона:

```sql
WHERE phone = '4801234567'        -- ложится на шкалу: 24 шага
WHERE address LIKE '%Ленина%'     -- не ложится: все десять миллионов строк
```

Разница не в индексе, а в форме запроса. Увидеть её заранее может тот, кто узнаёт в индексе `bin_first`.

Осталось посмотреть, почему на диске индекс устроен не так, как список в памяти.

Но таблица на диске это не список Python. Диск отдаёт данные кусками по несколько килобайт, такой кусок называют **страницей** (page). Одно чтение страницы с диска в тысячи раз медленнее, чем чтение из оперативной памяти. Бинарный поиск по отсортированному массиву на диске прыгал бы примерно по 15 разным страницам: первые 15 середин почти всегда на новой странице, и только последние девять шагов идут внутри одной.

Поэтому индекс устроен иначе. Номера клиентов разложены по страницам, скажем, по 500 номеров по порядку в каждой. Над ними страница-оглавление: она говорит, в какой из 500 нижних страниц искать. Над ней ещё одно оглавление. Трёх уровней по 500 записей хватает на `500 · 500 · 500 = 125` миллионов строк, и поиск читает три страницы вместо пятнадцати. Такую многоуровневую структуру называют **B-деревом** (B-tree).

**Предскажите.** Сколько уровней нужно для `10^10` строк при 500 записях на странице?

---

<details><summary>Ответ</summary>

Четыре: три уровня покрывают `125` миллионов, четыре уже `500^4 ≈ 6 · 10^10`. Поиск среди десяти миллиардов строк читает четыре страницы.
</details>

---

Аналогия: индекс это отсортированный массив, приспособленный под диск. Где аналогия врёт: индекс живёт отдельно от таблицы, и каждая вставка строки обновляет ещё и его. Индекс ускоряет чтение и замедляет запись. Индексы строят по тем столбцам, по которым ищут, а не по всем подряд.

Устройство B-дерева в этой лекции дальше не разбирается. Это всё ещё роль первая, только массив лежит на диске и стоит денег.

В следующей задаче списка не будет ни в памяти, ни на диске.

## 4. Роль вторая: поиск по ответу задачи

Та же игра, что в разделе 2, только массив никто не строит.

> **Задача D. Дипломы.** За школьные годы Ата накопил `n` дипломов с олимпиад, все одного размера: `w` в ширину и `h` в высоту. Теперь он учится в НИУ ВШЭ и хочет повесить их в комнате общежития на квадратную пробковую доску, которая занимает как можно меньше места на стене. Каждый диплом размещается строго в прямоугольнике `w` на `h`. Поворачивать дипломы на 90 градусов запрещается. Прямоугольники разных дипломов не должны иметь общих внутренних точек. Требуется вычислить минимальный размер стороны доски.
> **Ввод.** Три целых числа `w`, `h`, `n` (`1 ≤ w, h, n ≤ 10^9`).
> **Вывод.** Одно целое число, сторона доски.
> **Ограничения.** 1 секунда, 512 МБ.
>
> Примеры: `2 3 10 → 9`, `1 1 1 → 1`, `1 1 1000 → 32`.

Третий пример стоит проверить в уме: `31 × 31 = 961` дипломов мало, `32 × 32 = 1024` хватает. «Общих внутренних точек» значит: дипломы не перекрываются, а касаться краями могут. Поворот запрещён, поэтому ширина и высота не меняются местами.

Наивно: перебирать сторону `1, 2, 3, …`, пока дипломы не влезут. При `w = h = n = 10^9` дипломы образуют квадрат примерно `31623 × 31623` штук (`31623` это корень из `10^9`), и каждый шириной `10^9`. Сторона доски около `3 · 10^13`. При десяти миллионах шагов перебора в секунду это больше месяца.

**Попробуйте сами, прежде чем читать дальше.** Массива здесь нет. Что играет роль отсортированного массива? Какое свойство задачи заменяет сортировку?

---

<details><summary>Разбор</summary>

Возьмите доску со стороной `side`. В ряд по ширине помещается `side // w` дипломов, рядов по высоте `side // h`. Всего `(side // w) * (side // h)`. Проверка «влезает ли `n` штук» для каждой стороны даёт «нет» или «да».

Пройдите стороны по возрастанию: `нет, нет, нет, …, да, да, да`. Ни одного «да» перед «нет»: если влезло на доску `9`, влезет и на `10`. Это и есть отсортированный массив, только его никто не строит. Каждый его элемент вычисляется в момент, когда до него доходит середина.
</details>

---

Код с занятия:

````python
# введите «2 3 10»; ответ: 9
w, h, key = map(int, input().split())

left, right = 0, 10**18                 # 1. ГРАНИЦЫ: на left не влезает, на right влезает
while right - left > 1:                 # 2. ПОКА между ними есть целая сторона
    mid = (left + right) // 2           # 3. СЕРЕДИНА: кандидат в размер доски
    a = (mid // w) * (mid // h)         # 4. ПРОВЕРКА: сколько дипломов влезает на доску mid
    if a < key:                         #    меньше, чем нужно?
        left = mid                      #    да: ответ правее
    else:
        right = mid                     #    нет: ответ здесь или левее
print(right)                            # 5. ОТВЕТ: наименьшая сторона, на которую влезает
````

На занятии число дипломов назвали `key`: это порог, с которым сравнивают проверку, как искомое число в `bin_first`. Инвариант тот же по форме: на доску `left` не влезает, на доску `right` влезает. Границы выбраны так, чтобы он был верен без проверки. На доску со стороной `0` не влезает ничего. Правая граница взята по худшему случаю: даже в один ряд `10^9` дипломов шириной `10^9` занимают ровно `10^18`, и на такую доску влезает всё.

За первые 56 шагов `right` спустился от `10^18` до `13`, а `left` так и стоял на `0`: на всех этих досках дипломы влезали. Последние четыре шага для ввода `2 3 10`:

| left | right | mid | влезает | действие |
|---|---|---|---|---|
| 0 | 13 | 6 | `3 · 2 = 6` | `6 < 10`, `left = 6` |
| 6 | 13 | 9 | `4 · 3 = 12` | `right = 9` |
| 6 | 9 | 7 | `3 · 2 = 6` | `left = 7` |
| 7 | 9 | 8 | `4 · 2 = 8` | `left = 8` |
| 8 | 9 | | | стоп, ответ `9` |

**Предскажите.** Сколько всего раз выполнится тело цикла для ввода `2 3 10`? Что программа напечатает для ввода `1 1 10`?

---

<details><summary>Ответ</summary>

Для этого ввода 60 раз: `log2(10^18) ≈ 59.8`. Число шагов задаёт ширина начального отрезка, а не `w`, `h` и `n`. На других входах получается 59 или 60 (`1 1 10` даёт 59), смотря как округляются середины.

Для `1 1 10` ответ `4`: на доску `3 × 3` влезает `9` дипломов, на `4 × 4` уже `16`.
</details>

---

Сравните с `bin_first`:

| | `bin_first` | дипломы |
|---|---|---|
| что перебирается | индексы массива | размеры доски |
| откуда элемент | `arr[mid]` | считаем `(mid // w) * (mid // h)` |
| начальные границы | `-1` и `len(arr)` | `0` и `10^18` |
| ответ | `right` | `right` |

Скелет тот же. Этот приём называют **бинарный поиск по ответу** (binary search on the answer): шкалой служат возможные ответы задачи, здесь стороны доски. Вместо массива есть проверка «подходит ли кандидат». Сортировку заменяет монотонность: все «нет» левее всех «да».

### Как узнать бинарный поиск в новой задаче

Задача про бинарный поиск, если выполняются четыре признака:

1. есть шкала, на которой лежит ответ: индексы, размер, время, цена;
2. для одной точки шкалы можно дёшево проверить «подходит или нет»;
3. ответы проверки монотонны: сначала все «нет», потом все «да», или наоборот;
4. шкала большая, перебирать её подряд слишком долго.

Дальше всегда одно и то же: границы за краями шкалы, середина, сдвиг одной границы, пока границы не станут соседями.

**Проверьте признаки** на задаче `YES`/`NO` из раздела 2: что там шкала, что проверка, откуда монотонность?

---

<details><summary>Ответ</summary>

Шкала: индексы массива. Проверка: `arr[mid] <= key`, а не `arr[mid] == key`. Ответ «равно / не равно» не говорит, в какую сторону двигать границу, а «не больше ключа» слева даёт «да», справа «нет». Монотонность: массив отсортирован, поэтому все элементы не больше ключа стоят левее всех элементов больше ключа.
</details>

## 5. Роль третья: поиск корня уравнения

У дипломов сторона доски была целым числом, и среди целых есть наименьшее подходящее. Дальше шкала другая. Между любыми двумя вещественными числами есть третье, и наименьшего подходящего может не быть вовсе.

> **Задача B. Корень кубического уравнения.** Дано кубическое уравнение `a·x³ + b·x² + c·x + d = 0`, `a ≠ 0`. Известно, что у него ровно один корень. Требуется его найти.
> **Ввод.** Четыре целых числа через пробел, каждое от `−1000` до `1000`.
> **Вывод.** Единственный корень с точностью не менее 5 знаков после десятичной точки.
> **Ограничения.** 1 секунда, 64 МБ.

Примеры из контеста и то, что печатает код с занятия:

| ввод | ответ жюри | печатает код с занятия |
|---|---|---|
| `1 -3 3 -1` | `1.0000036491` | `0.9999953452376631` |
| `-1 -6 -12 -7` | `-1.0000000111` | `-0.9999999999886402` |

Судя по формулировке, проверяющая система принимает ответ, если он отличается от корня не больше чем на `10^-5`. Поэтому проходят все четыре числа. Корни здесь ровно `1` и `−1`: подставьте и проверьте. Почему в первой строке оба ответа промахиваются на несколько миллионных, объясняет последний подраздел раздела 6.

Уравнение `x³ = 2`, то есть ввод `1 0 0 -2`, имеет корень `1.2599…` с бесконечным хвостом цифр. Точного ответа в виде числа с конечной записью нет. Наивно: идти по отрезку от `−1001` до `1001` шагом `10^-5` и ловить смену знака. Откуда `1001`, чуть ниже. Это `2 · 10^8` вычислений `f`, около 40 секунд в Python при лимите в одну секунду.

**Попробуйте сами, прежде чем читать дальше.** Пройдите по четырём признакам из предыдущего раздела. Что здесь шкала, что проверка, откуда монотонность?

---

<details><summary>Разбор</summary>

Шкала: вещественные числа от `-10^9` до `10^9`. Проверка: знак `f(x)` в точке, где `f(x) = a·x³ + b·x² + c·x + d`.

Монотонность: у уравнения один корень, а знак `f` меняется только в корне. Значит, слева от корня знак один, справа другой, и смена ровно одна, как «нет» на «да» у дипломов. На концах отрезка знаки заведомо разные: при `x = ±10^9` слагаемое `a·x³` по модулю не меньше `10^27`, три остальных вместе не больше `10^21 + 10^12 + 10^3`, а знак `x³` слева и справа противоположный.
</details>

---

````python
# введите «1 0 0 -8»: корень ровно 2, а напечатается 2.0000000000043854
a, b, c, d = map(float, input().split())

f = lambda x: a * x ** 3 + b * x ** 2 + c * x + d   # 1. ФУНКЦИЯ, знак которой проверяем

eps = 10 ** -10                                     # 2. ТОЧНОСТЬ: при каком отрезке остановиться
left, right = -10 ** 9, 10 ** 9                     # 3. ГРАНИЦЫ: корень точно между ними
while right - left > eps:                           # 4. ПОКА отрезок длиннее eps
    mid = (left + right) / 2                        # 5. СЕРЕДИНА: деление вещественное
    if f(left) * f(mid) > 0:                        # 6. ПРОВЕРКА: на [left, mid] знак не сменился?
        left = mid                                  #    да: корень правее середины
    else:
        right = mid                                 #    нет: корень на [left, mid]
print((left + right) / 2)                           # 7. ОТВЕТ: середина последнего отрезка
````

Строка `f = lambda x: ...` это короткая запись функции из одного выражения. То же самое, что `def f(x): return a * x ** 3 + ...`.

Произведение `f(left) * f(mid)` положительно, когда знаки совпадают. Раз корень один, то на отрезке `[left, mid]` его нет, и левая граница прыгает в середину. Знак `f` слева от корня зависит от `a`: при `a = 1` там минус, а во втором примере контеста, где `a = −1`, плюс. Поэтому проверяют не `f(mid) < 0`, а совпадение со знаком в `left`.

Границы `±10^9` взяты с запасом в миллион раз: при коэффициентах от `−1000` до `1000` корень по модулю не больше `1001`. При `|x| ≥ 1001` слагаемое `a·x³` по модулю не меньше `1001 · x²`, а три остальных вместе меньше `1001 · x²` даже при коэффициентах `1000`. Значит, `f` там не ноль. Вне ограничений проверьте границы сами, иначе бисекция молча вернёт границу: для `1 0 0 -1e30` с корнем `10^10` код печатает `1000000000.0`.

Возьмите ввод `1 0 0 -8`: корень ровно `2`, код печатает `2.0000000000043854`. Первые два шага:

| left | right | mid | f(mid) | действие |
|---|---|---|---|---|
| `-10^9` | `10^9` | `0` | `-8` | `f(left) < 0` тоже, знаки совпали, `left = 0` |
| `0` | `10^9` | `5 · 10^8` | `≈ 1.25 · 10^26` | знак сменился, `right = 5 · 10^8` |
| … | … | … | … | ещё 63 шага |
| `1.99999999997728` | `2.00000000003149` | | | отрезок `5.4 · 10^-11 < eps`, стоп, ответ середина `2.0000000000043854` |

**Предскажите.** Сколько раз выполнится тело цикла? Отрезок длиной `2 · 10^9` нужно ужать до `10^-10`.

---

<details><summary>Ответ</summary>

`65` раз. После 64 делений отрезок равен `2 · 10^9 / 2^64 ≈ 1.08 · 10^-10`, а это ещё больше `eps`. Нужно 65-е деление. Как и у дипломов, число шагов зависит только от начального отрезка и точности.
</details>

---

Сравните с дипломами:

| | дипломы | корень уравнения |
|---|---|---|
| условие остановки | `right - left > 1` | `right - left > eps` |
| середина | `//`, целая | `/`, вещественная |
| проверка | `a < key` | `f(left) * f(mid) > 0` |
| ответ | `right`, ровно | середина отрезка, с погрешностью до `eps` |

Этот вариант бинарного поиска называют **методом бисекции** (bisection method). И это первый в лекции алгоритм, который не даёт точного ответа. Он даёт число, отличающееся от истинного корня меньше чем на `eps`. Такая гарантия имеет своё имя, и о нём в разделе 7. Но прежде чем давать имена, нужно разобраться с `eps`: почему условие остановки написано через него, а не через `f(mid) == 0`.

Три роли пройдены: массив в памяти, ответ задачи, корень уравнения. Скелет во всех трёх один. Менялось только то, откуда берётся элемент шкалы.

## 6. Почему eps, а не равенство

Почему в цикле выше не написать `while f(mid) != 0`? Ответ вскрывает маленький код с занятия. Он печатает число `0.1` в двоичной записи, цифра за цифрой: умножить на два, взять целую часть, отбросить её.

Почему это даёт двоичные цифры. Умножение на два сдвигает двоичную запятую, как умножение на десять сдвигает десятичную. Пример на `0.75`: `1.5` даёт цифру `1` и остаток `0.5`; `1.0` даёт цифру `1` и остаток `0`. Итого `0.11`, и правда `1/2 + 1/4 = 0.75`.

````python
x = 0.1

while x != 0:
    x *= 2
    print(int(x), end='')
    x -= int(x)
print()
````

**Предскажите.** Закончится ли этот цикл? Если да, сколько примерно цифр он напечатает? Запишите ответ. Ориентир: `0.1` в двоичной системе это периодическая дробь `0.0(0011)`, как `1/3 = 0.333…` в десятичной.

---

<details><summary>Ответ</summary>

Цикл заканчивается, напечатав 55 цифр:

```
0001100110011001100110011001100110011001100110011001101
```

Если вы ответили «бесконечный», вы рассуждали про математическое `0.1`. В переменной лежит не оно.
</details>

---

Три нуля в начале это не значащие цифры, а позиция: `0.1` меньше `1/8`, поэтому первые три двоичных разряда пустые. Уберите их и посчитайте остальное. На занятии это сделали так:

````python
print(len("1100110011001100110011001100110011001100110011001101"))
````

Печатает `52`. Тип `float` это 64 бита: один бит на знак, 11 бит на **порядок** (exponent), то есть на положение первой единицы, и 52 бита на **мантиссу** (mantissa), цифры после первой единицы. Сама первая единица не хранится: она есть у любого ненулевого числа. Значащих цифр поэтому 53. У `0.1` последняя из них ноль, и цикл его не печатает, отсюда 52. Проверьте на `x = 0.3`: цикл напечатает 54 цифры, первая из них ноль, значащих 53.

Дробь оборвалась там, где кончилась мантисса. Периодический хвост отрезан. Отброшенный хвост начинается с единицы, то есть он больше половины последнего оставленного разряда, и последнюю цифру округляют вверх, как `0.16` до `0.2`. Значащие 53 цифры обрываются на `…1100 1`, после прибавления единицы получается `…1101 0`. Последний ноль цикл не печатает, поэтому на экране `…1101` и 52 цифры.

В переменной `x` лежит не `0.1`, а ближайшая к нему дробь из 53 двоичных цифр. Модуль `Decimal` умеет печатать хранимое значение точно:

````python
from decimal import Decimal
print(Decimal(0.1))    # 0.1000000000000000055511151231257827021181583404541015625
````

Отсюда правило. Число, полученное вычислением, почти никогда не равно «красивому» значению в точности. Сравнение `f(mid) != 0` может не выполниться ни разу, и цикл не остановится. Поэтому вещественные алгоритмы останавливаются по `eps`: не «пока не попали», а «пока отрезок не стал достаточно коротким».

</details>

Диагноз ставится одной строкой внутри цикла: `print(left, right, mid)`. Вот та же бисекция с этой строкой и со счётчиком шагов, чтобы ячейка не висела вечно. В коде с занятия счётчика нет.

````python
a, b, c, d = map(float, input().split())      # введите: 1 0 0 -1e18

f = lambda x: a * x ** 3 + b * x ** 2 + c * x + d

eps = 10 ** -10
left, right = -10 ** 9, 10 ** 9
steps = 0
while right - left > eps:
    mid = (left + right) / 2
    steps += 1
    if steps > 66:
        print(left, right, mid)
    if steps > 70:
        print("остановлено вручную: в коде с занятия ограничителя нет")
        break
    if f(left) * f(mid) > 0:
        left = mid
    else:
        right = mid
````


Откуда берётся `1.16 · 10^-10`. Мантисса это 52 бита на всё число, и целая часть отбирает их у дробной. Этот раздел и два следующих, каждый про своё различие: какую гарантию даёт ответ, есть ли внутри случайность, как алгоритм исполняется. Различия независимы. Алгоритм получает место во всех трёх сразу, а не в одном.

**Точный алгоритм** (exact algorithm) возвращает **оптимум** (optimum), то есть наилучший возможный ответ, дальше `C_опт` (`C` от cost, стоимость или ценность решения). Бинарный поиск по массиву и по ответу точные: сторона `9` минимальна, и меньше не бывает.

**Приближённый алгоритм** (approximation algorithm) возвращает `C_прибл` с гарантией `|C_прибл − C_опт| < eps`: расстояние до оптимума, без учёта знака, меньше `eps`. Бисекция приближённая, и вы только что проверили эту гарантию на числах. Держится она, пока знак `f(mid)` вычислен верно, и раздел 6 показал, где это ломается.

### Рюкзак: одна задача, три подхода

Это **задача о рюкзаке** (knapsack problem). Есть `n` предметов, у каждого вес и ценность. Рюкзак выдерживает вес `W`. Нужно набрать предметы максимальной суммарной ценности, не превысив `W`. На занятии задачу разбирали без кода, поэтому здесь только идеи.

**Полный перебор** (brute force). Каждый предмет либо в рюкзаке, либо нет. Перебрать все наборы, посчитать ценность каждого, выбрать лучший. Точный.

**Предскажите.** Сколько наборов при `n = 20`? При `n = 60`?

---

<details><summary>Ответ</summary>

`2^20 ≈ 10^6` и `2^60 ≈ 10^18`. Миллион наборов компьютер перебирает за секунду. На `10^18` при миллиарде наборов в секунду уйдёт тридцать лет. Точный ответ есть, но добраться до него нельзя.
</details>

---

**Динамическое программирование** (dynamic programming). Строится таблица размером `n × W`: клетка `(i, j)` хранит лучшую ценность на первых `i` предметах при вместимости `j`. Считается она из двух соседних. Либо предмет `i` не в рюкзаке, и это клетка `(i − 1, j)`. Либо в рюкзаке, и это его ценность плюс клетка `(i − 1, j − вес_i)`. Из двух вариантов в клетку идёт больший. Тоже точный, работы `n · W` вместо `2^n`. При `n = 60` и `W = 1000` это 60 тысяч клеток.

**Предскажите.** А сколько клеток при `n = 60` и `W = 10^12`?

---

<details><summary>Ответ</summary>

`6 · 10^13`. Такая таблица не помещается в память. К тому же веса должны быть целыми, иначе столбцов не пронумеровать. Динамическое программирование точное, но не для любых `W`.
</details>

---

**Метод отжига** (simulated annealing). Алгоритм начинает с любого допустимого набора. Случайно меняет один предмет. Если ценность выросла, оставляет. Если упала, тоже иногда оставляет: иначе поиск застрянет в наборе, где любая одиночная замена ухудшает, а две замены подряд улучшили бы. Чем дальше идёт работа, тем реже принимаются ухудшения. Название от отжига металла: горячий металл перескакивает между состояниями свободно, остывший застывает. Где аналогия врёт: металл остывает сам, а здесь расписание «температуры» пишет программист, и остудить можно заново сколько угодно раз. Останавливается алгоритм по времени.

На задачах, где оптимум известен из перебора, отжиг обычно подходит к нему близко. Но для нового входа расстояние до оптимума неизвестно, и оценить его нельзя. Такой алгоритм называют **эвристическим** (heuristic). Он работает при любых `n` и `W`, и это его единственное обещание. Финальный проект дисциплины это попытка решить задачу именно таким алгоритмом.

**Предскажите.** Отжиг делает миллион замен в секунду и работает минуту. Какую долю из `2^60` наборов он успеет посмотреть?

---

<details><summary>Ответ</summary>

`6 · 10^7` из `10^18`, доля `6 · 10^-11`. Поэтому обещать оптимум он не может, и это не дефект реализации.
</details>

Одна задача, три алгоритма, три класса. Выбор класса это выбор между гарантией и временем.

### Проверка пароля: гарантия с вероятностью

Четвёртый класс возникает там, где под вопросом не время, а сам ответ. Пример с занятия.

Сайт должен проверять пароль при входе. Хранить пароли в базе нельзя: любая утечка базы станет утечкой всех паролей сразу. Поэтому хранят не пароль, а его **хеш** (hash), число фиксированной длины, которое считает **хеш-функция** (hash function). Функция устроена так, что по хешу восстановить пароль нельзя.

````python
import hashlib

print(hashlib.sha256("qwerty123".encode()).hexdigest())
# daaad6e5604e8e17bd9f108d91e26afe6281dac8fda0091040a7a6d7bd9b43b5
````

В базе лежит эта строка, а не `qwerty123`. При входе сайт хеширует введённое и сравнивает два хеша. Совпали, значит, пароль верный.

**Предскажите.** Хеш SHA-256 это 256 бит, то есть `2^256` разных значений. Паролей можно придумать сколько угодно. Может ли чужой пароль пройти проверку?

---

<details><summary>Ответ</summary>

Может. Паролей бесконечно много, а хешей конечное число, поэтому разные пароли обязательно где-то дают один и тот же хеш. Такое совпадение называют **коллизией** (collision). Сайт видит только равенство хешей и пустит любого, кто в коллизию попал.
</details>

---

Значит, ответ «пароль верный» не гарантирован, он верен с вероятностью. Вопрос в том, какой. Для двух наугад взятых паролей вероятность коллизии около `2^-256`, это `9 · 10^-78`. Для сравнения: если бы каждый житель Земли придумывал по миллиону паролей в секунду в течение миллиарда лет, вероятность хоть одного совпадения осталась бы около `3 · 10^-13`.

Алгоритм, который отвечает верно не всегда, а с вероятностью, называют **вероятностным** (probabilistic algorithm). Обратите внимание, чем он отличается от эвристики. Эвристика не может оценить, насколько её ответ далёк от лучшего. Здесь оценка есть, и она записана числом.

И это различие снова независимо от первых двух. Полный перебор и динамическое программирование оба точные и оба детерминированные, но перебор параллелится, а таблица динамического программирования нет: каждая клетка ждёт соседнюю.

| динамическое программирование | точный | детерминированный | последовательный |
| отжиг | эвристический | рандомизированный | параллелится по запускам |
| проверка пароля по хешу | вероятностный | детерминированный | последовательный |

Две клетки, где чаще всего ошибаются. Проверка пароля не бросает монетку, поэтому она детерминированная, хотя гарантия у неё вероятностная. Отжиг можно запустить несколько раз на разных ядрах и взять лучший из наборов, поэтому «параллелится по запускам»: один запуск последовательный, а запуски независимы.
</details>

## 10. Проверьте себя

Ответьте без кода, потом сверьтесь.

1. Массив из `10^6` элементов. Сколько шагов сделает `bin_first` в худшем случае?
2. В задаче о дипломах `left = 0`. Какой инвариант должен быть верен перед циклом, и почему его не нужно проверять?
3. Объясните своими словами, почему цикл `while x != 0` для `x = 0.1` закончился.
4. Отжиг для рюкзака вернул ценность `1000`, а точный ответ `1040`. Почему это не ошибка в программе?
5. Что изменится в бисекции, если у уравнения три корня? Найдёт ли она хотя бы один?

<details><summary>Ответы</summary>

1. Двадцать: `log2(10^6) ≈ 19.9`, округляем вверх.
2. «На доску со стороной `left` дипломы не влезают». На доску со стороной `0` не влезает ничего, это верно без вычислений.
3. В переменной лежит не `0.1`, а двоичная дробь из 53 значащих цифр. У конечной дроби конечное число цифр, и после последней `x` становится нулём.
4. Эвристика не обещает оптимум. Она обещает «обычно хороший» ответ за отведённое время. Ошибкой было бы выдать набор тяжелее `W`.

Хуже, когда корень не меняет знак: у `1 -4 5 -2` корни `1` и `2`, но в единице `f` касается нуля, не пересекая его: `f = (x − 1)² · (x − 2)`, множитель `(x − 1)²` не бывает отрицательным, а `(x − 2)` по обе стороны от единицы отрицателен. Бисекция находит только `2`, а единицу не найдёт никогда. Условие «ровно один корень» в задаче не украшение.

## 11. Итог

- Бинарный поиск отбрасывает половину: `n` вариантов за `log2(n)` шагов, миллиард за тридцать.
- Массив не обязателен. Нужны шкала, дешёвая проверка и монотонность: «нет» до «да». Три роли одного приёма: массив в памяти, ответ задачи, корень уравнения.
- Сравнивайте вещественные числа через `eps`, и берите `eps` крупнее шага `float` в районе ответа.
- Алгоритмы различают по гарантии на ответ: точный, приближённый, эвристический, вероятностный. Отдельно по случайности внутри и по параллельности. Финальный проект дисциплины это эвристика.

**Где встретится дальше.** Сортировка, без которой бинарный поиск не работает. Деревья, в том числе B-дерево из раздела 3. Хеш-функции и хеш-таблицы. Динамическое программирование для рюкзака. Эвристики в финальном проекте.
